```python
#Q.19]
def pythagorean_triplets(N):
    for a in range(1, N + 1):
        for b in range(a, N + 1):
            for c in range(b, N + 1):
                if a*a + b*b == c*c:
                    print((a, b, c))

N = int(input("Enter N: "))
pythagorean_triplets(N)
```
